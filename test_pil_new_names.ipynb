```python

path = "Desktop"

def get_subdirs(path : str):
    dirs = [d for d in os.listdir(path) if os.path.isdir(os.path.join(path, d))]
    return dirs

def get_current_files(path : str):
    current_files = [f for f in os.listdir(path) if os.path.isfile(os.path.join(path, f))]
    return current_files

## Consider to put all_paths here
all_paths = []

def get_all_subdirectories_with_files(path : str, all_paths):
    num_subdirs = 1
    if len(get_current_files(path)) > 0:
        all_paths.append(path)
    current_subdirs = get_subdirs(path)
    for current_subdir in current_subdirs:
        new_path = os.path.join(path, current_subdir)
        get_all_subdirectories_with_files(new_path, all_paths)

get_all_subdirectories_with_files("Desktop", all_paths)
print(f"Paths are as: \t {all_paths}")
```
