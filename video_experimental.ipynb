```python
# Libarary Imports

from pymediainfo import MediaInfo
import os


# Local Imports
from sort_images import get_all_subdirectories_with_files, get_fnames_from_path, get_suffix
```

```python
movies_suffixes = [".MOV", ".mov"]
print(os.listdir())

ignore_folders = ['.venv', '.ipynb_checkpoints', 'Pictures_With_Dates', 'test_pil_new_names.ipynb', 'Pictures_Potential_Duplicates', 'Pictures_Without_Date', '.git', 'some_experimentation.ipynb',
'__pycache__', 'sort_videos.py', 'sort_images.py', '.gitignore', 'video_experimental.ipynb']

all_paths = []
my_path = os.getcwd()
print(my_path)
```

```python
get_all_subdirectories_with_files(my_path, all_paths, ignore_folders)
```

```python
print(len(all_paths))
```

```python
movie_count = 0
for i, path in enumerate(all_paths):
    fnames = get_fnames_from_path(path)
    
    for fname in fnames:
        
        current_suffix = get_suffix(fname)
        if current_suffix in movies_suffixes:
            movie_count += 1

print(f"There is a total of:\t {movie_count} .MOV-files")
```

```python
test_path = os.path.join(all_paths[500],get_fnames_from_path(all_paths[500])[0])
print(test_path)
```

```python
media_info = MediaInfo.parse(test_path)
general_track = media_info.general_tracks[0]

# Sjekk de vanligste datofeltene (returnerer None hvis de ikke finnes)
encoded_date = general_track.encoded_date
tagged_date = general_track.tagged_date
file_modified_date = general_track.file_last_modification_date

# Skriv ut resultatene
print(f"Innspilt/Kodethistorikk: {encoded_date}")
print(f"Tagget dato: {tagged_date}")
print(f"Filendringsdato: {file_modified_date}")
```
