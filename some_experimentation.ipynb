```python
base_name = potential_duplicate_file_list[0]
base_name_no_suffix = base_name.strip(get_suffix(base_name))

dictionary_overview = {}
list_of_duplicates = []

for i, next_fname in enumerate(potential_duplicate_file_list):
    
    if base_name_no_suffix in next_fname:
        list_of_duplicates.append(next_fname)
    else:
        dictionary_overview[base_name_no_suffix] = list_of_duplicates
        base_name = next_fname
        base_name_no_suffix = base_name.strip(get_suffix(base_name))
        list_of_duplicates = [next_fname]
```

```python
count_keys = 0
count_items = 0
for key in dictionary_overview:
    items = len(dictionary_overview[key])
    count_keys += 1
    count_items += items
    print(f"{key} with {items}")
print(count_items / count_keys)
```

```python
from PIL import Image
from PIL.ExifTags import TAGS

def print_meta_data_of_image(img_path):
    # open the image
    image = Image.open(img_path)
    
    # extracting the exif metadata
    exifdata = image.getexif()
    
    # looping through all the tags present in exifdata
    for tagid in exifdata:
        
        # getting the tag name instead of tag id
        tagname = TAGS.get(tagid, tagid)
    
        # passing the tagid to get its respective value
        value = exifdata.get(tagid)
      
        # printing the final result
        print(f"{tagname:25}: {value}")
```

```python
key = "2013_07_19_18_51_58"
print(len(dictionary_overview[key]))

for i in range(len(dictionary_overview[key])):
    img_path = os.path.join("Pictures_Potential_Duplicates", dictionary_overview[key][i])
    print_meta_data_of_image(img_path)
```
